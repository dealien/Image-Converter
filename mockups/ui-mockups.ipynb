# Image Converter | UI Mockups
### ⚠️ Important: Run the Initialization cell below first!
This notebook leverages actual production code (`rich_menu.py`) while bypassing interactive blocking inputs using mocks.

```python
import sys
import os
from unittest.mock import MagicMock
import questionary
from rich.console import Console

try:
    from rich.group import Group
except ImportError:
    # Fallback for certain environments
    from rich.console import Group

# 1. Ensure project root is in path
project_root = os.path.abspath("..")
if project_root not in sys.path:
    sys.path.insert(0, project_root)

import rich_menu  # noqa: E402

console = Console()


# 2. Define Mocks
def mock_metadata(path):
    basename = os.path.basename(path)
    data = {
        "Hill Castle.png": ("1920 × 1080", "2.4 MB", "PNG"),
        "Spacecraft.png": ("1280 × 720", "1.8 MB", "PNG"),
        "Tree Clear Sky 1.png": ("2560 × 1440", "3.1 MB", "PNG"),
        "Tree Clear Sky 2.png": ("2560 × 1440", "2.9 MB", "PNG"),
    }
    return data.get(basename, ("1024 × 1024", "1.0 MB", "JPG"))


def mock_checkbox(message, choices, **kwargs):
    lines = []
    for i, choice in enumerate(choices):
        icon = "[bold green]☑[/]" if i < 2 else "[ ]"
        lines.append(f"{icon} {choice.title}")
    if "instruction" in kwargs:
        lines.append(f"\n[dim]{kwargs['instruction']}[/]")
    console.print(Group(*lines))
    m = MagicMock()
    m.ask.return_value = [c.value for c in choices[:2]]
    return m


# 3. Apply Mocks
rich_menu._get_image_metadata = mock_metadata
rich_menu.console.clear = lambda: None  # Prevent clearing notebook output
questionary.checkbox = mock_checkbox

print("✅ Environment initialized, project modules linked, and mocks applied.")
```
